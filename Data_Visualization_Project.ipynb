# Data Visualization Using Python

## Introduction

Data Visualization is the process of representing data
using charts and graphs.

In this project, we use Python libraries such as
Matplotlib, Seaborn, and Pandas to visualize data.

## Objectives

1. Create different charts and graphs.
2. Understand patterns and trends in data.
3. Analyze sales and profit information.
4. Generate useful insights from data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
              'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],

    'Sales': [12000, 15000, 18000, 14000, 20000, 22000,
              25000, 23000, 27000, 30000, 28000, 35000],

    'Profit': [2000, 2500, 3000, 2200, 3500, 4000,
               4500, 4200, 5000, 6000, 5500, 7000],

    'Product': ['Mobile', 'Laptop', 'Tablet', 'Mobile',
                'Laptop', 'Tablet', 'Mobile', 'Laptop',
                'Tablet', 'Mobile', 'Laptop', 'Tablet'],

    'Region': ['North', 'South', 'East', 'West',
              'North', 'South', 'East', 'West',
              'North', 'South', 'East', 'West']
}

df = pd.DataFrame(data)

print(df)
```

```python
print("First 5 Rows:")
display(df.head())

print("Dataset Information:")
df.info()

print("Statistical Summary:")
display(df.describe())
```

```python
product_sales = df.groupby('Product')['Sales'].sum()

plt.figure(figsize=(8,5))

plt.bar(product_sales.index, product_sales.values)

plt.title("Total Sales by Product")
plt.xlabel("Products")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Month'], df['Sales'],
         marker='o', linestyle='-')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(7,7))

plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Sales Distribution by Region")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Sales'], bins=6, edgecolor='black')

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Sales'], df['Profit'])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=df, x='Product', y='Profit')

plt.title("Average Profit by Product")
plt.xlabel("Product")
plt.ylabel("Average Profit")

plt.show()
```

```python
plt.figure(figsize=(6,5))

correlation = df[['Sales', 'Profit']].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Sales and Profit Correlation")

plt.show()
```

```python
print("DATA ANALYSIS REPORT")

print("----------------------------")

print("Total Sales:", df['Sales'].sum())

print("Total Profit:", df['Profit'].sum())

print("Average Sales:", df['Sales'].mean())

print("Highest Sales:",
      df.loc[df['Sales'].idxmax(), 'Month'])

print("Highest Profit:",
      df.loc[df['Profit'].idxmax(), 'Month'])

print("Best Selling Product:",
      product_sales.idxmax())

print("Best Region:",
      region_sales.idxmax())
```
